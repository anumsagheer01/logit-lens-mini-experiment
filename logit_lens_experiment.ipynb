```python
import torch
from transformer_lens import HookedTransformer
```

```python
def run_logit_lens(model, prompt, use_chat_template=True):
    if use_chat_template:
        messages = [{"role": "user", "content": prompt}]
        text = model.tokenizer.apply_chat_template(
            messages, tokenize=False, add_generation_prompt=True
        )
    else:
        text = prompt

    tokens = model.to_tokens(text)
    logits, cache = model.run_with_cache(tokens)
    last_pos = tokens.shape[1] - 1

    results = {}
    for layer in range(model.cfg.n_layers):
        resid = cache["resid_post", layer][0, last_pos]
        resid_normed = model.ln_final(resid)
        layer_logits = resid_normed @ model.W_U
        top5 = torch.topk(layer_logits, 5).indices
        results[layer] = [model.to_string(t) for t in top5]

    return results
```

```python
model_gpt2 = HookedTransformer.from_pretrained("gpt2")

prompt_raw = "Concentrate on the concept of oranges. Now copy this sentence exactly: The old painting hung crookedly on the"
results_gpt2 = run_logit_lens(model_gpt2, prompt_raw, use_chat_template=False)

for layer, tokens in results_gpt2.items():
    print(f"Layer {layer}: {tokens}")
```

```python
del model_gpt2
torch.cuda.empty_cache()
```

```python
model_qwen_small = HookedTransformer.from_pretrained("Qwen/Qwen1.5-0.5B-Chat")

prompt = "Concentrate on the concept of oranges. Now copy this sentence exactly: The old painting hung crookedly on the"
results_qwen_small = run_logit_lens(model_qwen_small, prompt, use_chat_template=True)

for layer, tokens in results_qwen_small.items():
    print(f"Layer {layer}: {tokens}")
```
